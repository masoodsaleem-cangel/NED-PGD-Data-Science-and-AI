```python
patients = [] #keeping it outside, becaause Everytime loop repeats it resets patients to Zero
while True: # C++ Equalent of while (true) {....} A self Start
  print("\nWhat would you like to do?")
  print("1. Press 'yes' to add a new patient")
  print("2. Press 'p' to print all patients")
  print("3. Press 'q' to quit")

  choice = input("\nYour choice: ").lower() #incase case sensitivity bothers

  if choice == 'yes' or choice == 'y':
    print("\n--- Add New Patient ---")
    name = input("Enter Patient Name: ")
    patient_id = input("Enter Patient ID: ")
    blood_type = input("Enter Blood Type (A+, A-, B+, B-, O+, O-, AB+, AB-): ")
    phone = input("Enter Phone Number: ")
    # put everthing in tuple below
    patient = (name, patient_id, blood_type, phone)
    # Now add tuple to list defined at the start of Program
    patients.append(patient)
    print(f"\n✓ Patient {name} added successfully!")
  elif choice == 'p':
    if len(patients) == 0: #incase someone try to print with zero patient
      print("\n--- No patients in record ---")
      print("Please add some patients first!")
    else:
      print("\n" + "="*50) # Makeup
      print("ALL PATIENTS LIST")
      print("="*50) #Makeup
      for i, patient in enumerate(patients, 1): #Count and label
        print(f"\nPatient #{i}")
        print(f"Name: {patient[0]}")
        print(f"ID: {patient[1]}")
        print(f"Blood Type: {patient[2]}")
        print(f"Phone: {patient[3]}")
        print("-"*30)
        print(f"Total Patients: {len(patients)}")
  elif choice == 'q':
      print("\nThank you for using Patient Record System!")
      break # Break is associated with elif above, and elif with while, it will break while too, if we put outside it, it will break after every choice
  else: # this else is child of while, Any other choice it will print below and will repeat the while loop
    print("\nInvalid choice! Please enter 'yes', 'p', or 'q'")
```
